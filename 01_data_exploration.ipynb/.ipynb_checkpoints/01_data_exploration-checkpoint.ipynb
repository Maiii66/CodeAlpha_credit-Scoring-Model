```python
# ============================================================================
# 📋 COPY THIS ENTIRE CODE INTO YOUR JUPYTER NOTEBOOK
# ============================================================================
# Just copy everything below and paste into a Jupyter cell
# Press Shift+Enter to run
# ============================================================================

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# YOUR FILE PATH (CORRECTED)
file_path = r'C:\Users\maiyarasu\Downloads\statlog+german+credit+data\german.data'

# COLUMN NAMES FOR GERMAN CREDIT (21 columns)
column_names = [
    'Status_of_existing_checking_account',
    'Duration_in_months',
    'Credit_history',
    'Purpose',
    'Credit_amount',
    'Savings_account_bonds',
    'Present_employment_since',
    'Installment_rate_as_percent_of_disposable_income',
    'Personal_status_and_sex',
    'Other_debtors',
    'Present_residence_since',
    'Property',
    'Age_in_years',
    'Other_installment_plans',
    'Housing',
    'Number_of_existing_credits',
    'Job',
    'Number_of_people_being_liable',
    'Telephone',
    'Foreign_worker',
    'class'
]

# LOAD THE DATA
df = pd.read_csv(
    file_path,
    header=None,
    names=column_names,
    sep='\s+',
    engine='python'
)

# DISPLAY RESULTS
print("=" * 70)
print("✅ DATASET LOADED SUCCESSFULLY!")
print("=" * 70)

print(f"\nDataset Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"Memory Usage: {df.memory_usage().sum() / 1024**2:.2f} MB")

print("\n" + "=" * 70)
print("FIRST 5 ROWS:")
print("=" * 70)
print(df.head())

print("\n" + "=" * 70)
print("DATA TYPES:")
print("=" * 70)
print(df.dtypes)

print("\n" + "=" * 70)
print("MISSING VALUES:")
print("=" * 70)
missing = df.isnull().sum()
print(f"Total missing values: {missing.sum()}")
if missing.sum() == 0:
    print("✓ NO MISSING VALUES - Data is clean!")

print("\n" + "=" * 70)
print("BASIC STATISTICS:")
print("=" * 70)
print(df.describe())

print("\n" + "=" * 70)
print("TARGET VARIABLE (class):")
print("=" * 70)
print(df['class'].value_counts())
print("\nClass Meaning:")
print("  1 = Good Credit (Approved)")
print("  2 = Bad Credit (Rejected)")

# SAVE TO YOUR PROJECT FOLDER
output_path = r'C:\Users\maiyarasu\CodeAlpha_CreditScoringModel\data\raw_data.csv'
df.to_csv(output_path, index=False)
print(f"\n✅ Data saved to: {output_path}")

print("\n" + "=" * 70)
print("✅ DATA LOADING COMPLETE - READY FOR EXPLORATION!")
print("=" * 70)
```
