```python
# ============================================================================
# DATA LOADING & INSPECTION - German Credit Dataset
# Portable: all paths are relative to the project folder, so this runs fine
# on this machine, on a server, or in Colab (as long as data/raw_data.csv exists).
# ============================================================================

import pandas as pd
import numpy as np
import warnings
from pathlib import Path

warnings.filterwarnings('ignore')

# Locate project root (this notebook lives in its own subfolder)
if Path.cwd().name == '01_data_exploration.ipynb':
    project_root = Path.cwd().parent
else:
    project_root = Path.cwd()

data_path = project_root / 'data' / 'raw_data.csv'

if not data_path.exists():
    raise FileNotFoundError(
        f'Data file not found: {data_path}\n'
        'Ensure raw_data.csv exists in the data/ folder (it was generated from '
        'the original Statlog german.data file).'
    )

# LOAD THE DATA (raw_data.csv already contains the 21 named columns)
df = pd.read_csv(data_path)

# DISPLAY RESULTS
print("=" * 70)
print("DATASET LOADED SUCCESSFULLY!")
print("=" * 70)

print(f"\nDataset Shape: {df.shape[0]} rows x {df.shape[1]} columns")
print(f"Memory Usage: {df.memory_usage().sum() / 1024**2:.2f} MB")

print("\n" + "=" * 70)
print("FIRST 5 ROWS:")
print("=" * 70)
print(df.head())

print("\n" + "=" * 70)
print("DATA TYPES:")
print("=" * 70)
print(df.dtypes.astype(str))

print("\n" + "=" * 70)
print("MISSING VALUES:")
print("=" * 70)
missing = df.isnull().sum()
print(f"Total missing values: {missing.sum()}")
if missing.sum() == 0:
    print("NO MISSING VALUES - Data is clean!")

print("\n" + "=" * 70)
print("BASIC STATISTICS:")
print("=" * 70)
print(df.describe())

print("\n" + "=" * 70)
print("TARGET VARIABLE (class):")
print("=" * 70)
print(df['class'].value_counts())
print("\nClass Meaning:")
print("  1 = Good Credit (Approved)")
print("  2 = Bad Credit (Rejected)")
```
